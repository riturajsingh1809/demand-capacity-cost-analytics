```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load cleaned dataset
file_path = "Cleaned_Data.xlsx"
df = pd.read_excel(file_path)

# Convert date
df["Date"] = pd.to_datetime(df["Date"])

# Basic information
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nSummary Statistics:")
print(df.describe())

# ---------------------------------------------------------
# Daily Demand Analysis
# ---------------------------------------------------------

daily = (
    df.groupby("Date")
      .agg(
          Traffic=("Traffic", "sum"),
          Applications=("Applications", "sum"),
          CPU_Units=("CPU_Units", "sum"),
          Storage_GB=("Storage_GB", "sum"),
          Infrastructure_Cost=("Infrastructure_Cost", "sum")
      )
      .reset_index()
      .sort_values("Date")
)

print("\nDaily Demand Summary:")
print(daily.head())

# ---------------------------------------------------------
# Monthly Demand Analysis
# ---------------------------------------------------------

daily["Month"] = daily["Date"].dt.to_period("M").astype(str)

monthly = (
    daily.groupby("Month")
         .agg(
             Traffic=("Traffic", "sum"),
             Applications=("Applications", "sum"),
             CPU_Units=("CPU_Units", "sum"),
             Storage_GB=("Storage_GB", "sum"),
             Infrastructure_Cost=("Infrastructure_Cost", "sum")
         )
         .reset_index()
)

monthly["Date"] = pd.to_datetime(monthly["Month"])

print("\nMonthly Demand:")
print(monthly.head())

# ---------------------------------------------------------
# Monthly Growth
# ---------------------------------------------------------

monthly["Traffic_Growth_%"] = monthly["Traffic"].pct_change() * 100
monthly["Application_Growth_%"] = monthly["Applications"].pct_change() * 100

# ---------------------------------------------------------
# Peak Demand Analysis
# ---------------------------------------------------------

peak_summary = (
    df.groupby("Peak_Status")
      .agg(
          Total_Traffic=("Traffic", "sum"),
          Total_Applications=("Applications", "sum"),
          Average_Traffic=("Traffic", "mean"),
          Average_Applications=("Applications", "mean"),
          Average_Cost=("Infrastructure_Cost", "mean")
      )
      .reset_index()
)

print("\nPeak Demand Summary:")
print(peak_summary)

# ---------------------------------------------------------
# Placement Season Analysis
# ---------------------------------------------------------

placement_months = [5, 6, 7, 8]

placement_data = df[df["Date"].dt.month.isin(placement_months)]

normal_data = df[~df["Date"].dt.month.isin(placement_months)]

placement_avg_traffic = placement_data["Traffic"].mean()
normal_avg_traffic = normal_data["Traffic"].mean()

placement_avg_applications = placement_data["Applications"].mean()
normal_avg_applications = normal_data["Applications"].mean()

print("\nPlacement Season Analysis:")
print("Placement Season Average Traffic:", round(placement_avg_traffic, 2))
print("Normal Season Average Traffic:", round(normal_avg_traffic, 2))
print("Placement Season Average Applications:", round(placement_avg_applications, 2))
print("Normal Season Average Applications:", round(normal_avg_applications, 2))

# ---------------------------------------------------------
# Peak Traffic
# ---------------------------------------------------------

peak_traffic = daily["Traffic"].max()
peak_traffic_date = daily.loc[daily["Traffic"].idxmax(), "Date"]

peak_applications = daily["Applications"].max()
peak_application_date = daily.loc[daily["Applications"].idxmax(), "Date"]

print("\nPeak Traffic:", peak_traffic)
print("Peak Traffic Date:", peak_traffic_date.date())

print("\nPeak Applications:", peak_applications)
print("Peak Application Date:", peak_application_date.date())

# ---------------------------------------------------------
# Recent Demand Trend
# ---------------------------------------------------------

recent_months = monthly.tail(6)

recent_traffic_avg = recent_months["Traffic"].mean()
previous_traffic_avg = monthly.iloc[-12:-6]["Traffic"].mean()

traffic_change = (
    (recent_traffic_avg - previous_traffic_avg)
    / previous_traffic_avg
) * 100

print("\nRecent Traffic Change:", round(traffic_change, 2), "%")

# ---------------------------------------------------------
# Simple Moving Average Forecast
# ---------------------------------------------------------

monthly["Traffic_MA_3"] = monthly["Traffic"].rolling(3).mean()
monthly["Applications_MA_3"] = monthly["Applications"].rolling(3).mean()

traffic_forecast = monthly["Traffic"].tail(3).mean()
application_forecast = monthly["Applications"].tail(3).mean()

print("\nNext Period Traffic Forecast:", round(traffic_forecast, 2))
print("Next Period Applications Forecast:", round(application_forecast, 2))

# ---------------------------------------------------------
# Backtesting
# ---------------------------------------------------------

backtest = monthly.copy()

backtest["Forecast_Traffic"] = backtest["Traffic"].shift(1).rolling(3).mean()
backtest["Forecast_Applications"] = backtest["Applications"].shift(1).rolling(3).mean()

backtest = backtest.dropna()

traffic_mae = np.mean(
    np.abs(
        backtest["Traffic"] - backtest["Forecast_Traffic"]
    )
)

application_mae = np.mean(
    np.abs(
        backtest["Applications"] - backtest["Forecast_Applications"]
    )
)

print("\nBacktest Results:")
print("Traffic MAE:", round(traffic_mae, 2))
print("Applications MAE:", round(application_mae, 2))

# ---------------------------------------------------------
# Capacity Projection
# ---------------------------------------------------------

current_traffic = traffic_forecast
current_applications = application_forecast

current_cpu = monthly["CPU_Units"].tail(3).mean()
current_storage = monthly["Storage_GB"].tail(3).mean()
current_cost = monthly["Infrastructure_Cost"].tail(3).mean()

scales = [2, 5, 10]

capacity_projection = pd.DataFrame({
    "Scale": scales,
    "Projected_Traffic": [current_traffic * x for x in scales],
    "Projected_Applications": [current_applications * x for x in scales],
    "Projected_CPU_Units": [current_cpu * x for x in scales],
    "Projected_Storage_GB": [current_storage * x for x in scales],
    "Projected_Infrastructure_Cost": [current_cost * x for x in scales]
})

print("\nCapacity and Cost Projection:")
print(capacity_projection.round(2))

# ---------------------------------------------------------
# Cost per Traffic Unit
# ---------------------------------------------------------

cost_per_traffic = (
    monthly["Infrastructure_Cost"].sum()
    / monthly["Traffic"].sum()
)

cost_per_application = (
    monthly["Infrastructure_Cost"].sum()
    / monthly["Applications"].sum()
)

print("\nCost per Traffic Unit:", round(cost_per_traffic, 4))
print("Cost per Application:", round(cost_per_application, 4))

# ---------------------------------------------------------
# Regional Analysis
# ---------------------------------------------------------

regional = (
    df.groupby("Region")
      .agg(
          Traffic=("Traffic", "sum"),
          Applications=("Applications", "sum"),
          Average_CPU=("CPU_Units", "mean"),
          Average_Cost=("Infrastructure_Cost", "mean")
      )
      .reset_index()
)

print("\nRegional Analysis:")
print(regional.sort_values("Traffic", ascending=False))

# ---------------------------------------------------------
# Platform Analysis
# ---------------------------------------------------------

platform_analysis = (
    df.groupby("Platform")
      .agg(
          Traffic=("Traffic", "sum"),
          Applications=("Applications", "sum"),
          Average_CPU=("CPU_Units", "mean"),
          Average_Cost=("Infrastructure_Cost", "mean")
      )
      .reset_index()
)

print("\nPlatform Analysis:")
print(platform_analysis.sort_values("Traffic", ascending=False))

# ---------------------------------------------------------
# Key Business Metrics
# ---------------------------------------------------------

total_traffic = df["Traffic"].sum()
total_applications = df["Applications"].sum()
total_cost = df["Infrastructure_Cost"].sum()

average_daily_traffic = daily["Traffic"].mean()
average_daily_applications = daily["Applications"].mean()

print("\nKey Business Metrics:")
print("Total Traffic:", total_traffic)
print("Total Applications:", total_applications)
print("Total Infrastructure Cost:", round(total_cost, 2))
print("Average Daily Traffic:", round(average_daily_traffic, 2))
print("Average Daily Applications:", round(average_daily_applications, 2))

# ---------------------------------------------------------
# Risk and Assumption Indicators
# ---------------------------------------------------------

traffic_std = monthly["Traffic"].std()
traffic_cv = traffic_std / monthly["Traffic"].mean()

print("\nDemand Variability:")
print("Traffic Standard Deviation:", round(traffic_std, 2))
print("Traffic Coefficient of Variation:", round(traffic_cv, 4))

if traffic_cv < 0.20:
    variability = "Low"
elif traffic_cv < 0.40:
    variability = "Moderate"
else:
    variability = "High"

print("Demand Variability Level:", variability)

# ---------------------------------------------------------
# Forecast Summary
# ---------------------------------------------------------

forecast_summary = pd.DataFrame({
    "Metric": [
        "Forecast Traffic",
        "Forecast Applications",
        "Peak Traffic",
        "Peak Applications",
        "Traffic Backtest MAE",
        "Applications Backtest MAE",
        "Placement Season Avg Traffic",
        "Normal Season Avg Traffic",
        "Cost per Traffic Unit",
        "Cost per Application"
    ],
    "Value": [
        round(traffic_forecast, 2),
        round(application_forecast, 2),
        peak_traffic,
        peak_applications,
        round(traffic_mae, 2),
        round(application_mae, 2),
        round(placement_avg_traffic, 2),
        round(normal_avg_traffic, 2),
        round(cost_per_traffic, 4),
        round(cost_per_application, 4)
    ]
})

print("\nForecast Summary:")
print(forecast_summary)

# ---------------------------------------------------------
# Save One Analysis Result File
# ---------------------------------------------------------

output_file = "Python Analysis Result.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    monthly.to_excel(writer, sheet_name="Monthly Analysis", index=False)
    peak_summary.to_excel(writer, sheet_name="Peak Analysis", index=False)
    capacity_projection.to_excel(writer, sheet_name="Capacity Projection", index=False)
    regional.to_excel(writer, sheet_name="Regional Analysis", index=False)
    platform_analysis.to_excel(writer, sheet_name="Platform Analysis", index=False)
    forecast_summary.to_excel(writer, sheet_name="Forecast Summary", index=False)

print("\nAnalysis completed successfully.")
print("Output file:", output_file)
```
